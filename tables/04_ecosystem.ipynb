```r
safe_name <- function(x) {
    x %>% str_replace_all("[ /&:;,+()\\[\\]<U+2013><U+2014>]+", "_") %>% str_replace_all("_+", "_") %>% str_replace_all("^_|_$", "")
}
clean_metabolite_name <- function(x) {
    z <- as.character(x)
    z <- stringr::str_replace_all(z, "\\s*\\([^\\)]*\\)", "")
    stringr::str_squish(z)
}
get_metab_source <- function(x) {
    case_when(str_detect(x, "\\(B\\)$") ~ "Blood", str_detect(x, "\\(U\\)$") ~ "Urine", TRUE ~ "Unknown")
}
compact_middle_label <- function(x, max_chars = 40) {
    z <- stringr::str_squish(as.character(x))
    z[is.na(z)] <- ""
    out <- vapply(z, function(s) {
        n <- nchar(s)
        if (!is.finite(n) || n <= max_chars) {
            return(s)
        }
        tail_n <- min(10L, max(6L, floor(max_chars * 0.24)))
        head_n <- max_chars - tail_n - 1L
        paste0(substr(s, 1L, head_n), "<U+2026>", substr(s, n - tail_n + 1L, n))
    }, character(1))
    unname(out)
}
short_metabolite_label <- function(x, source = NULL) {
    raw <- as.character(x)
    if (is.null(source)) {
        src <- get_metab_source(raw)
    }
    else {
        src <- as.character(source)
    }
    base <- clean_metabolite_name(raw)
    suffix <- dplyr::case_when(src == "Blood" ~ " (B)", src == "Urine" ~ " (U)", TRUE ~ "")
    paste0(compact_middle_label(base, max_chars = METAB_LABEL_MAX_CHARS), suffix)
}
BUBBLE_COLUMN_LABEL_ROT <- 35
EXTERNAL_ONLY_BOTTOM_COLUMN_LABELS <- TRUE
METAB_SHOW_MODULE_TITLES_MAIN <- FALSE
METAB_SHOW_MODULE_TITLES_EXTERNAL <- FALSE
FINAL15C_SHOW_ROW_LABELS <- FALSE
FINAL15C_SHOW_AXIS_TITLES <- FALSE
FINAL15C_COLUMN_NAMES_ROT <- 32
METABOLIC_MODULE_LEGEND_LABELS <- c(`Urea-cycle metabolism` = "Arginine\n<U+2192> Ornithine / urea cycle", `BCAA metabolism` = "Leucine / Isoleucine / Valine\n<U+2192> Branched-chain metabolism", `PUFA metabolism` = "Arachidonic acid / DHA\n<U+2192> Lipid mediators", `TCA cycle` = "cis-Aconitate\n<U+2192> TCA-cycle metabolism", `Aromatic co-metabolism` = "Phenylalanine / phenethylacetate\n<U+2192> Aromatic co-metabolites", `Tryptophan metabolism` = "Tryptophan\n<U+2192> Kynurenine", `SCFA metabolism` = "Microbial SCFAs\n<U+2192> Host energy metabolism", 
    `Amino-acid metabolism` = "Histidine / Lysine\n<U+2192> Amino-acid derivatives", `Fatty-acid metabolism` = "Fatty acids\n<U+2192> Lipid mediators", `Purine metabolism` = "Inosine\n<U+2192> Hypoxanthine / purines", `Methyl-donor metabolism` = "Betaine\n<U+2192> One-carbon cycle & methylation", `Butyrate metabolism` = "Tributyrin\n<U+2192> Butyrate", `Acylcarnitine metabolism` = "Acylcarnitines\n<U+2192> Fatty-acid oxidation", `Acetate metabolism` = "Acetate\n<U+2192> Acetyl-CoA", `Propionate metabolism` = "Propionate\n<U+2192> Propionyl-CoA", 
    `Membrane lipid metabolism` = "Phospholipids\n<U+2192> Membrane remodeling", `Other axis-linked metabolism` = "Metabolites\n<U+2192> Axis-linked metabolism")
METAB_LEGEND_ROW_SPACING <- 1.22
METAB_LEGEND_BAR_HALF_HEIGHT <- 0.46
METAB_LEGEND_TEXT_SIZE <- 3.35
METAB_LEGEND_LINEHEIGHT <- 0.88
METAB_LEGEND_WIDTH_IN <- 5.4
METAB_LEGEND_ROW_HEIGHT_IN <- 0.29
METAB_LEGEND_NCOL <- 2
add_suffix_if_missing <- function(mat, suffix) {
    rn <- rownames(mat)
    has_suffix <- str_detect(rn, "\\s*\\([BU]\\)$")
    rownames(mat) <- ifelse(has_suffix, rn, paste0(rn, " ", suffix))
    mat
}
normalize_key <- function(x) {
    x <- as.character(x)
    x <- str_trim(x)
    x <- str_replace_all(x, "\\s*\\([BU]\\)$", "")
    x <- str_to_lower(x)
    x <- str_replace_all(x, "[^a-z0-9]+", "")
    x
}
p_to_star <- function(p) {
    ifelse(is.na(p), "", ifelse(p < 0.001, "***", ifelse(p < 0.01, "**", ifelse(p < 0.05, "*", ""))))
}
safe_cor_test <- function(x, y) {
    ok <- is.finite(x) & is.finite(y)
    if (sum(ok) < 5) 
        return(c(rho = NA_real_, p = NA_real_, n = sum(ok)))
    z <- tryCatch(suppressWarnings(cor.test(x[ok], y[ok], method = "spearman", exact = FALSE)), error = function(e) NULL)
    if (is.null(z)) 
        return(c(rho = NA_real_, p = NA_real_, n = sum(ok)))
    c(rho = unname(z$estimate), p = z$p.value, n = sum(ok))
}
z_vec <- function(x) {
    x <- as.numeric(x)
    if (sum(is.finite(x)) < 3 || !is.finite(sd(x, na.rm = TRUE)) || sd(x, na.rm = TRUE) == 0) {
        return(rep(NA_real_, length(x)))
    }
    as.numeric(scale(x))
}
read_any_csv <- function(file) {
    if (!file.exists(file)) 
        stop("Missing file: ", file)
    as.data.frame(data.table::fread(file, data.table = FALSE, check.names = FALSE), check.names = FALSE)
}
set_first_col_as_rownames <- function(x) {
    if (ncol(x) < 2) 
        return(x)
    v <- x[[1]]
    nm <- colnames(x)[1]
    looks_id <- length(unique(v)) == nrow(x) && (is.character(v) || is.factor(v) || tolower(nm) %in% c("sample", "sample_id", "patient", "patient_id", "id", "feature", "ko", "name"))
    if (looks_id) {
        rownames(x) <- as.character(v)
        x[[1]] <- NULL
    }
    x
}
read_matrix_auto <- function(file, sample_ids, want = c("feature_by_sample", "sample_by_feature")) {
    want <- match.arg(want)
    x <- read_any_csv(file)
    sid <- toupper(trimws(as.character(sample_ids)))
    if (ncol(x) >= 2) {
        first_ids <- toupper(trimws(as.character(x[[1]])))
        if (sum(first_ids %in% sid, na.rm = TRUE) >= 2 && length(unique(first_ids)) == nrow(x)) {
            rownames(x) <- first_ids
            x[[1]] <- NULL
        }
        else {
            x <- set_first_col_as_rownames(x)
        }
    }
    rn_up <- toupper(trimws(rownames(x)))
    cn_up <- toupper(trimws(colnames(x)))
    row_hit <- sum(rn_up %in% sid)
    col_hit <- sum(cn_up %in% sid)
    if (want == "feature_by_sample") {
        if (row_hit > col_hit) {
            rownames(x) <- rn_up
            x <- as.data.frame(t(x), check.names = FALSE)
        }
        else {
            colnames(x) <- cn_up
        }
    }
    else {
        if (col_hit > row_hit) {
            colnames(x) <- cn_up
            x <- as.data.frame(t(x), check.names = FALSE)
        }
        else {
            rownames(x) <- rn_up
        }
    }
    x[] <- lapply(x, function(v) suppressWarnings(as.numeric(v)))
    x <- as.data.frame(x, check.names = FALSE)
    if (want == "feature_by_sample") {
        colnames(x) <- toupper(trimws(colnames(x)))
    }
    else {
        rownames(x) <- toupper(trimws(rownames(x)))
    }
    x
}
rbind_feature_matrices <- function(...) {
    mats <- list(...)
    all_samples <- Reduce(union, lapply(mats, colnames))
    mats2 <- lapply(mats, function(m) {
        miss <- setdiff(all_samples, colnames(m))
        if (length(miss) > 0) {
            for (s in miss) m[[s]] <- NA_real_
        }
        m[, all_samples, drop = FALSE]
    })
    do.call(rbind, mats2)
}
```

```r
normalize_search_text <- function(x) {
    x <- tolower(ifelse(is.na(x), "", as.character(x)))
    x <- str_replace_all(x, "[^a-z0-9]+", " ")
    x <- str_replace_all(x, "\\s+", " ")
    trimws(x)
}
contains_phrase_boundary <- function(text, phrase) {
    text_n <- paste0(" ", normalize_search_text(text), " ")
    phrase_n <- normalize_search_text(phrase)
    str_detect(text_n, fixed(paste0(" ", phrase_n, " ")))
}
contains_any_phrase_boundary <- function(text, phrases) {
    if (length(phrases) == 0) 
        return(rep(FALSE, length(text)))
    hit <- rep(FALSE, length(text))
    for (ph in phrases) hit <- hit | contains_phrase_boundary(text, ph)
    hit
}
extract_ko_vector <- function(x) {
    toupper(str_extract(as.character(x), "K\\d{5}"))
}
detect_ko_column <- function(df) {
    scores <- vapply(df, function(v) sum(!is.na(extract_ko_vector(v))), numeric(1))
    if (max(scores, na.rm = TRUE) == 0) {
        stop("Cannot detect KO IDs in KEGG annotation. Columns: ", paste(colnames(df), collapse = ", "))
    }
    names(scores)[which.max(scores)]
}
kegg_anno_raw <- read_any_csv(KEGG_ANNOTATION_FILE)
ko_col_hcc <- detect_ko_column(kegg_anno_raw)
kegg_anno_raw$ko_id <- extract_ko_vector(kegg_anno_raw[[ko_col_hcc]])
kegg_anno_raw <- kegg_anno_raw %>% filter(!is.na(ko_id))
text_cols <- setdiff(colnames(kegg_anno_raw), c(ko_col_hcc, "ko_id"))
text_cols <- text_cols[vapply(kegg_anno_raw[text_cols], function(v) {
    is.character(v) || is.factor(v)
}, logical(1))]
if (length(text_cols) == 0) {
    stop("No textual fields are available in the HCC KEGG annotation for keyword search.")
}
annotation_text <- do.call(paste, c(lapply(kegg_anno_raw[text_cols], function(v) ifelse(is.na(v), "", as.character(v))), sep = " | "))
biomap_decode <- NULL
if (file.exists(BIOMAP_KEGG_DECODE)) {
    biomap_decode <- read_any_csv(BIOMAP_KEGG_DECODE)
    if (ncol(biomap_decode) >= 2) {
        first_ko <- extract_ko_vector(biomap_decode[[1]])
        if (sum(!is.na(first_ko)) > 0) {
            biomap_decode$ko_id <- first_ko
        }
        else {
            kk <- detect_ko_column(biomap_decode)
            biomap_decode$ko_id <- extract_ko_vector(biomap_decode[[kk]])
        }
    }
}
ko_description_col <- detect_col(kegg_anno_raw, c("Description", "description", "KO Description", "KO_description", "ko_description", "Definition", "definition", "Gene description", "gene_description", "KEGG description", "KEGG_description", "Function", "function"), required = FALSE)
if (is.null(ko_description_col)) {
    warning("No KEGG description column detected. ", "KO display labels will fall back to gene symbol / KO ID.")
} else {
}
extract_ko_description <- function(ko_ids) {
    out <- rep(NA_character_, length(ko_ids))
    if (!is.null(ko_description_col)) {
        mm <- match(ko_ids, kegg_anno_raw$ko_id)
        z <- as.character(kegg_anno_raw[[ko_description_col]][mm])
        z <- stringr::str_replace(z, stringr::regex("^K\\d{5}\\s*[:|\\-]?\\s*", ignore_case = TRUE), "")
        z <- stringr::str_replace_all(z, "\\s*\\[EC:[^\\]]+\\]", "")
        z <- stringr::str_split_fixed(z, ";", 2)[, 1]
        z <- stringr::str_squish(z)
        good <- !is.na(z) & z != ""
        out[good] <- z[good]
    }
    out
}
make_ko_display_label <- function(ko_ids, descriptions, gene_symbols) {
    full <- as.character(descriptions)
    gene <- as.character(gene_symbols)
    use_gene <- is.na(full) | full == ""
    full[use_gene] <- gene[use_gene]
    use_id <- is.na(full) | full == ""
    full[use_id] <- ko_ids[use_id]
    compact_middle_label(full, max_chars = KO_LABEL_MAX_CHARS)
}
if (!file.exists(KO_DESCRIPTION_FILE)) {
    stop("KO description file not found: ", KO_DESCRIPTION_FILE)
}
ko_description_reference <- read_any_csv(KO_DESCRIPTION_FILE)
if (!all(c("gene_id", "Description") %in% colnames(ko_description_reference))) {
    stop(get_value("science.messages.status_030", "input must contain columns: "), "gene_id and Description. Found: ", paste(colnames(ko_description_reference), collapse = ", "))
}
ko_description_reference <- ko_description_reference %>% transmute(ko_id = stringr::str_extract(toupper(as.character(gene_id)), "K\\d{5}"), Description = as.character(Description)) %>% filter(!is.na(ko_id)) %>% distinct(ko_id, .keep_all = TRUE)
pick_preferred_gene_alias <- function(x) {
    z <- as.character(x)
    vapply(z, function(s) {
        if (is.na(s) || stringr::str_trim(s) == "") {
            return(NA_character_)
        }
        s <- stringr::str_squish(s)
        s <- stringr::str_replace(s, stringr::regex("^K\\d{5}\\s*[:|=\\-]?\\s*", ignore_case = TRUE), "")
        primary <- stringr::str_split_fixed(s, ";", 2)[, 1]
        primary <- stringr::str_trim(primary)
        parts <- unlist(stringr::str_split(primary, "\\s*[,/|]\\s*"))
        parts <- stringr::str_trim(parts)
        parts <- parts[parts != ""]
        parts <- parts[!stringr::str_detect(parts, "^K\\d{5}$")]
        expanded <- character(0)
        for (p in parts) {
            if (stringr::str_detect(p, "^[A-Za-z][A-Za-z0-9._-]{1,14}$")) {
                expanded <- c(expanded, p)
            }
            else {
                lead <- stringr::str_match(p, "^([A-Za-z][A-Za-z0-9._-]{1,14})(?:\\s+|$)")[, 2]
                if (!is.na(lead)) 
                  expanded <- c(expanded, lead)
            }
        }
        token_like <- unique(expanded)
        if (length(token_like) == 0) {
            return(NA_character_)
        }
        bacterial_like <- token_like[stringr::str_detect(token_like, "^[a-z][A-Za-z0-9._-]*$")]
        if (length(bacterial_like) > 0) {
            return(bacterial_like[1])
        }
        mixed_like <- token_like[stringr::str_detect(token_like, "[a-z]") & stringr::str_detect(token_like, "[A-Z]")]
        if (length(mixed_like) > 0) {
            return(mixed_like[1])
        }
        token_like[1]
    }, character(1), USE.NAMES = FALSE)
}
extract_gene_symbol <- function(ko_ids) {
    mm <- match(ko_ids, ko_description_reference$ko_id)
    desc_raw <- as.character(ko_description_reference$Description[mm])
    alias <- pick_preferred_gene_alias(desc_raw)
    alias
}
make_ko_alias_display <- function(ko_ids, gene_symbols) {
    gene <- as.character(gene_symbols)
    gene[is.na(gene) | gene == ""] <- NA_character_
    gene
}
pathway_rows <- list()
audit_rows <- list()
counter <- 1
for (axis_i in AXIS_LEVELS) {
    subdefs <- AXIS_DEFINITION[[axis_i]]
    for (sub_i in names(subdefs)) {
        kws <- subdefs[[sub_i]]
        hit <- contains_any_phrase_boundary(annotation_text, kws)
        if (!any(hit)) {
            warning("No KEGG annotation hit for: ", axis_i, " | ", sub_i)
            next
        }
        tmp <- kegg_anno_raw[hit, , drop = FALSE]
        tmp <- tmp %>% distinct(ko_id, .keep_all = TRUE) %>% filter(ko_id %in% rownames(kegg_in))
        if (nrow(tmp) == 0) {
            warning("KEGG hits found but none present in HCC KO matrix for: ", axis_i, " | ", sub_i)
            next
        }
        matched_text <- annotation_text[hit][match(tmp$ko_id, kegg_anno_raw$ko_id[hit])]
        matched_kw <- vapply(matched_text, function(txt) {
            z <- kws[vapply(kws, function(k) contains_phrase_boundary(txt, k), logical(1))]
            paste(z, collapse = "; ")
        }, character(1))
        ko_gene_symbol <- extract_gene_symbol(tmp$ko_id)
        ko_description_full <- extract_ko_description(tmp$ko_id)
        pathway_rows[[counter]] <- data.frame(axis = axis_i, subpathway_en = sub_i, ko_id = tmp$ko_id, gene_symbol = ko_gene_symbol, ko_alias_display = make_ko_alias_display(tmp$ko_id, ko_gene_symbol), ko_description_full = ko_description_full, ko_description_display = make_ko_display_label(tmp$ko_id, ko_description_full, ko_gene_symbol), matched_keywords = matched_kw, stringsAsFactors = FALSE)
        counter <- counter + 1
    }
}
pathway_anno <- bind_rows(pathway_rows)
if (nrow(pathway_anno) == 0) {
    stop("No pathway-related KO overlaps the HCC KO matrix.")
}
ko_multisubpath_audit <- pathway_anno %>% count(axis, ko_id, name = "n_subpathway_hits") %>% filter(n_subpathway_hits > 1)
write.csv(ko_multisubpath_audit, file.path(dirs$tables, get_value("science.tables.result_245", "result_245.csv")), row.names = FALSE)
subpath_order_tbl <- bind_rows(lapply(AXIS_LEVELS, function(ax) {
    data.frame(axis = ax, subpathway_en = names(AXIS_DEFINITION[[ax]]), subpath_order = seq_along(AXIS_DEFINITION[[ax]]), stringsAsFactors = FALSE)
}))
pathway_anno <- pathway_anno %>% left_join(subpath_order_tbl, by = c("axis", "subpathway_en")) %>% arrange(match(axis, AXIS_LEVELS), subpath_order, ko_id) %>% group_by(axis, ko_id) %>% slice(1) %>% ungroup() %>% select(-subpath_order)
calc_two_group_feature_stats <- function(mat, features, group_a, group_b) {
    features <- unique(features[features %in% rownames(mat)])
    ga <- intersect(group_a, colnames(mat))
    gb <- intersect(group_b, colnames(mat))
    if (length(ga) < MIN_GROUP_N || length(gb) < MIN_GROUP_N) {
        stop("Too few samples for KO R-vs-NR admission: R=", length(ga), " NR=", length(gb))
    }
    bind_rows(lapply(features, function(feat) {
        x <- as.numeric(mat[feat, ga, drop = TRUE])
        y <- as.numeric(mat[feat, gb, drop = TRUE])
        x <- x[is.finite(x)]
        y <- y[is.finite(y)]
        pooled <- c(x, y)
        pooled_sd <- sd(pooled, na.rm = TRUE)
        effect_z <- if (length(x) >= MIN_GROUP_N && length(y) >= MIN_GROUP_N && is.finite(pooled_sd) && pooled_sd > 0) {
            (mean(x, na.rm = TRUE) - mean(y, na.rm = TRUE))/pooled_sd
        }
        else {
            NA_real_
        }
        p <- tryCatch(suppressWarnings(wilcox.test(x, y, exact = FALSE)$p.value), error = function(e) NA_real_)
        data.frame(ko_id = feat, n_responder = length(x), n_nonresponder = length(y), response_effect_z = effect_z, response_p = p, stringsAsFactors = FALSE)
    }))
}
ko_response_stats <- calc_two_group_feature_stats(kegg_in, unique(pathway_anno$ko_id), response_groups$Responder, response_groups$`Non-Responder`)
pathway_anno_all <- pathway_anno %>% left_join(ko_response_stats, by = "ko_id") %>% group_by(axis, subpathway_en) %>% mutate(response_q_subpathway = p.adjust(response_p, method = "BH"), ko_pass_fdr = is.finite(response_q_subpathway) & response_q_subpathway < KO_FDR_CUTOFF, ko_pass_effect = is.finite(response_effect_z) & abs(response_effect_z) >= KO_EFFECT_CUTOFF, ko_eligible = ko_pass_fdr & ko_pass_effect) %>% ungroup()
write.csv(pathway_anno_all, file.path(dirs$tables, get_value("science.tables.result_246", "result_246.csv")), row.names = FALSE)
write.csv(pathway_anno_all %>% filter(ko_eligible), file.path(dirs$tables, get_value("science.tables.result_247", "result_247.csv")), row.names = FALSE)
pathway_anno <- pathway_anno_all %>% filter(ko_eligible)
if (nrow(pathway_anno) == 0) {
    stop("No KO passed the prespecified KO significance/effect thresholds.")
}
```

```r
calc_spearman_edge_table <- function(anchor_mat, target_mat, samples, min_group_n = MIN_GROUP_N) {
    common <- Reduce(intersect, list(samples, colnames(anchor_mat), colnames(target_mat)))
    if (length(common) < min_group_n) {
        return(data.frame())
    }
    a <- as.matrix(anchor_mat[, common, drop = FALSE])
    b <- as.matrix(target_mat[, common, drop = FALSE])
    storage.mode(a) <- "double"
    storage.mode(b) <- "double"
    if (nrow(a) == 0 || nrow(b) == 0) 
        return(data.frame())
    cor_mat <- suppressWarnings(cor(x = t(a), y = t(b), method = "spearman", use = "pairwise.complete.obs"))
    if (is.null(dim(cor_mat))) {
        cor_mat <- matrix(cor_mat, nrow = nrow(a), ncol = nrow(b), dimnames = list(rownames(a), rownames(b)))
    }
    rownames(cor_mat) <- rownames(a)
    colnames(cor_mat) <- rownames(b)
    n_mat <- (is.finite(a) * 1) %*% t(is.finite(b) * 1)
    dimnames(n_mat) <- dimnames(cor_mat)
    rho_for_p <- pmax(pmin(cor_mat, 1 - 1e-12), -1 + 1e-12)
    df_mat <- pmax(n_mat - 2, 1)
    t_mat <- rho_for_p * sqrt(df_mat/pmax(1 - rho_for_p^2, .Machine$double.eps))
    p_mat <- 2 * stats::pt(-abs(t_mat), df = df_mat)
    p_mat[!is.finite(cor_mat) | n_mat < min_group_n] <- NA_real_
    p_mat[is.finite(cor_mat) & abs(cor_mat) >= 1 & n_mat >= min_group_n] <- 0
    dimnames(p_mat) <- dimnames(cor_mat)
    out <- expand.grid(anchor = rownames(cor_mat), target = colnames(cor_mat), stringsAsFactors = FALSE)
    out$n_pair <- as.vector(n_mat)
    out$rho <- as.vector(cor_mat)
    out$p_value <- as.vector(p_mat)
    out
}
bh_adjust_safe <- function(p) {
    out <- rep(NA_real_, length(p))
    ok <- is.finite(p)
    if (any(ok)) 
        out[ok] <- p.adjust(p[ok], method = "BH")
    out
}
calc_response_stats_generic <- function(mat, features, group_r, group_nr) {
    features <- unique(features[features %in% rownames(mat)])
    if (length(features) == 0) 
        return(data.frame())
    rr <- calc_two_group_feature_stats(mat, features, group_r, group_nr)
    rr %>% rename(feature = ko_id, response_effect_z = response_effect_z, response_p = response_p)
}
summarize_significant_targets <- function(edge_tbl) {
    if (nrow(edge_tbl) == 0) 
        return(data.frame())
    edge_tbl %>% group_by(target) %>% summarise(biomap_score_max = abs(sum(rho[significant], na.rm = TRUE)), biomap_score_mean = if (any(significant)) 
        mean(abs(rho[significant]), na.rm = TRUE)
    else 0, n_significant_links_max = sum(significant, na.rm = TRUE), min_edge_q = if (any(significant & is.finite(q_value))) {
        min(q_value[significant & is.finite(q_value)], na.rm = TRUE)
    }
    else NA_real_, max_abs_rho = if (any(significant & is.finite(rho))) {
        max(abs(rho[significant & is.finite(rho)]), na.rm = TRUE)
    }
    else NA_real_, .groups = "drop") %>% filter(n_significant_links_max > 0) %>% arrange(desc(n_significant_links_max), desc(biomap_score_max), min_edge_q, desc(max_abs_rho)) %>% mutate(best_rank = row_number(), selected_groups = "All")
}
parse_radiomics_family <- function(x) {
    case_when(str_detect(x, regex("(^|_)shape(_|$)", ignore_case = TRUE)) ~ "Morphology", str_detect(x, regex("(^|_)firstorder(_|$)", ignore_case = TRUE)) ~ "First-order", str_detect(x, regex("(^|_)glcm(_|$)", ignore_case = TRUE)) ~ "GLCM texture", str_detect(x, regex("(^|_)glrlm(_|$)", ignore_case = TRUE)) ~ "GLRLM texture", str_detect(x, regex("(^|_)glszm(_|$)", ignore_case = TRUE)) ~ "GLSZM texture", str_detect(x, regex("(^|_)gldm(_|$)", ignore_case = TRUE)) ~ "GLDM texture", str_detect(x, regex("(^|_)ngtdm(_|$)", 
        ignore_case = TRUE)) ~ "NGTDM texture", TRUE ~ "Other radiomics")
}
parse_radiomics_filter <- function(x) {
    case_when(str_detect(x, regex("^original", ignore_case = TRUE)) ~ "Original", str_detect(x, regex("wavelet", ignore_case = TRUE)) ~ "Wavelet", str_detect(x, regex("log|sigma", ignore_case = TRUE)) ~ "LoG", TRUE ~ "Other")
}
```

```r
normalize_metabolite_key <- function(x) {
    z <- as.character(x)
    z[is.na(z)] <- ""
    z <- str_replace_all(z, "&#39;", "'")
    z <- str_replace(z, "\\s*\\([BU]\\)$", "")
    z <- str_replace(z, "\\.[0-9]+$", "")
    z <- str_replace(z, "_[0-9.]+%$", "")
    z <- str_replace(z, "(?i)_major$", "")
    z <- str_replace(z, "(?i)\\s*\\(not validated[^)]*\\)", "")
    z <- str_replace(z, "(?i)\\s*\\(isomer of[^)]*\\)", "")
    z <- str_replace(z, "\\s*;.*$", "")
    z <- str_to_lower(trimws(z))
    z <- str_replace_all(z, "[^a-z0-9]+", "")
    z
}
read_metabolite_whitelist_one <- function(file) {
    if (!file.exists(file)) 
        stop("Metabolite annotation file not found: ", file)
    sheets <- readxl::excel_sheets(file)
    sheet_use <- if ("Selected_all" %in% sheets) 
        "Selected_all"
    else sheets[1]
    x <- suppressMessages(readxl::read_excel(file, sheet = sheet_use))
    src <- basename(file)
    if ("Standardized_name" %in% colnames(x)) {
        base <- x %>% transmute(standardized_name = as.character(.data[["Standardized_name"]]), category = as.character(.data[["Category"]]), subtype = if ("Subtype" %in% colnames(x)) 
            as.character(.data[["Subtype"]])
        else NA_character_, confidence = if ("Confidence" %in% colnames(x)) 
            as.character(.data[["Confidence"]])
        else NA_character_, rationale = if ("Rationale" %in% colnames(x)) 
            as.character(.data[["Rationale"]])
        else NA_character_, original_variants = if ("Original_variants" %in% colnames(x)) 
            as.character(.data[["Original_variants"]])
        else NA_character_)
        primary <- base %>% transmute(standardized_name, category, subtype, confidence, rationale, match_name = standardized_name, annotation_file = src)
        variants <- base %>% filter(!is.na(original_variants), original_variants != "") %>% transmute(standardized_name, category, subtype, confidence, rationale, match_name = original_variants, annotation_file = src) %>% separate_rows(match_name, sep = "\\s*\\|\\s*")
        out <- bind_rows(primary, variants)
    }
    else if (all(c("Original name", "Cleaned base name") %in% colnames(x))) {
        base <- x %>% transmute(standardized_name = as.character(.data[["Cleaned base name"]]), original_name = as.character(.data[["Original name"]]), category = as.character(.data[["Category"]]), subtype = if ("Subtype" %in% colnames(x)) 
            as.character(.data[["Subtype"]])
        else NA_character_, confidence = if ("Confidence" %in% colnames(x)) 
            as.character(.data[["Confidence"]])
        else NA_character_, rationale = if ("Rationale" %in% colnames(x)) 
            as.character(.data[["Rationale"]])
        else NA_character_)
        primary <- base %>% transmute(standardized_name, category, subtype, confidence, rationale, match_name = standardized_name, annotation_file = src)
        original <- base %>% transmute(standardized_name, category, subtype, confidence, rationale, match_name = original_name, annotation_file = src)
        out <- bind_rows(primary, original)
    }
    else {
        stop("Unrecognized metabolite annotation layout in: ", file, "\nColumns: ", paste(colnames(x), collapse = ", "))
    }
    out %>% mutate(match_name = trimws(as.character(match_name)), match_key = normalize_metabolite_key(match_name), category = trimws(as.character(category)), subtype = trimws(as.character(subtype)), confidence = trimws(as.character(confidence))) %>% filter(match_key != "", category %in% METAB_ALLOWED_CATEGORIES)
}
whitelist_all <- bind_rows(lapply(METAB_ANNO_FILES, read_metabolite_whitelist_one))
whitelist_unique <- whitelist_all %>% mutate(confidence_score = case_when(str_to_lower(confidence) == "high" ~ 3L, str_to_lower(confidence) == "medium" ~ 2L, str_to_lower(confidence) == "low" ~ 1L, TRUE ~ 0L)) %>% arrange(match_key, desc(confidence_score)) %>% group_by(match_key) %>% slice(1) %>% ungroup()
build_metabolite_catalog <- function(feature_names, source_name) {
    data.frame(metabolite = feature_names, metabolite_clean = clean_metabolite_name(feature_names), source = source_name, stringsAsFactors = FALSE) %>% mutate(match_key = normalize_metabolite_key(metabolite_clean)) %>% left_join(whitelist_unique %>% select(match_key, standardized_name, category, subtype, confidence, rationale, annotation_file), by = "match_key") %>% mutate(eligible = !is.na(standardized_name) & category %in% METAB_ALLOWED_CATEGORIES, subpathway_en = case_when(!is.na(subtype) & subtype != 
        "" ~ subtype, !is.na(category) & category != "" ~ category, TRUE ~ paste0(source, " metabolites")))
}
blood_catalog <- build_metabolite_catalog(rownames(blood_in), "Blood")
urine_catalog <- build_metabolite_catalog(rownames(urine_in), "Urine")
metab_catalog <- bind_rows(blood_catalog, urine_catalog)
blood_pool <- blood_catalog$metabolite[blood_catalog$eligible]
urine_pool <- urine_catalog$metabolite[urine_catalog$eligible]
write.csv(metab_catalog, file.path(dirs$tables, get_value("science.tables.result_248", "result_248.csv")), row.names = FALSE)
```

```r
feature_group_effect <- function(mat, group_a, group_b) {
    features <- rownames(mat)
    out <- lapply(features, function(feat) {
        x <- as.numeric(mat[feat, group_a[group_a %in% colnames(mat)], drop = TRUE])
        y <- as.numeric(mat[feat, group_b[group_b %in% colnames(mat)], drop = TRUE])
        allv <- c(x, y)
        zv <- z_vec(allv)
        zx <- zv[seq_along(x)]
        zy <- zv[length(x) + seq_along(y)]
        effect <- mean(zx, na.rm = TRUE) - mean(zy, na.rm = TRUE)
        p <- tryCatch(suppressWarnings(wilcox.test(x[is.finite(x)], y[is.finite(y)])$p.value), error = function(e) NA_real_)
        data.frame(feature = feat, effect = effect, p = p, stringsAsFactors = FALSE)
    })
    bind_rows(out) %>% mutate(star = p_to_star(p))
}
if (!is.finite(BUBBLE_EFFECT_LIM)) {
    bubble_effect_reference <- c(pathway_anno$response_effect_z, selected_mets$response_effect_z, selected_rad$response_effect_z)
    bubble_effect_reference <- abs(bubble_effect_reference[is.finite(bubble_effect_reference)])
    if (length(bubble_effect_reference) == 0) {
        BUBBLE_EFFECT_LIM <- 1
    }
    else {
        BUBBLE_EFFECT_LIM <- as.numeric(stats::quantile(bubble_effect_reference, probs = 0.95, na.rm = TRUE, names = FALSE))
        BUBBLE_EFFECT_LIM <- max(BUBBLE_EFFECT_LIM, 0.5)
    }
}
plot_bubble_one_comparison <- function(feature_mat, feature_tbl, feature_col, label_col, module_col, module_colors, group_a_samples, group_b_samples, comparison_label, outfile, axis_title = NULL, row_side = "right", body_col_width_mm = BUBBLE_COL_WIDTH_MM, row_label_max_width_mm = ROW_LABEL_MAX_WIDTH_MM, annotation_width_mm = 3, pdf_width_in = BUBBLE_SINGLE_PDF_WIDTH_IN, show_module_titles = FALSE, show_column_names = TRUE, column_names_rot = BUBBLE_COLUMN_LABEL_ROT) {
    feats <- unique(feature_tbl[[feature_col]])
    feats <- feats[feats %in% rownames(feature_mat)]
    if (length(feats) == 0) 
        return(NULL)
    ft <- feature_tbl[match(feats, feature_tbl[[feature_col]]), , drop = FALSE]
    st <- feature_group_effect(feature_mat[feats, , drop = FALSE], group_a_samples, group_b_samples)
    st <- st[match(feats, st$feature), , drop = FALSE]
    mat <- matrix(st$effect, ncol = 1, dimnames = list(feats, comparison_label))
    star <- matrix(st$star, ncol = 1, dimnames = list(feats, comparison_label))
    lim <- BUBBLE_EFFECT_LIM
    col_fun <- circlize::colorRamp2(c(-lim, 0, lim), c("#3C5488", "white", "#E64B35"))
    module_vec <- as.character(ft[[module_col]])
    module_vec[is.na(module_vec) | module_vec == ""] <- "Unclassified"
    module_display_vec <- if (isTRUE(show_module_titles)) {
        ko_module_display_label(module_vec)
    }
    else {
        module_vec
    }
    module_factor <- factor(module_display_vec, levels = unique(module_display_vec))
    anno <- rowAnnotation(Module = anno_simple(module_vec, col = module_colors, width = unit(annotation_width_mm, "mm")), show_annotation_name = FALSE, show_legend = FALSE)
    labels <- as.character(ft[[label_col]])
    labels[is.na(labels) | labels == ""] <- feats[is.na(labels) | labels == ""]
    ht <- Heatmap(mat, name = "Effect", col = col_fun, show_heatmap_legend = FALSE, width = grid::unit(ncol(mat) * body_col_width_mm, "mm"), cluster_rows = FALSE, cluster_columns = FALSE, row_split = module_factor, row_title = if (isTRUE(show_module_titles)) 
        "%s"
    else NULL, row_title_rot = 0, row_title_side = "left", row_title_gp = gpar(fontsize = if (isTRUE(show_module_titles)) 
        KO_MODULE_TITLE_FONT_PT
    else 0, fontfamily = PLOT_FONT_FAMILY, fontface = "bold"), row_gap = grid::unit(0.65, "mm"), show_row_names = TRUE, row_labels = labels, row_names_side = row_side, row_names_gp = gpar(fontsize = 7.8, fontfamily = PLOT_FONT_FAMILY), row_names_max_width = grid::unit(row_label_max_width_mm, "mm"), show_column_names = show_column_names, column_names_rot = column_names_rot, column_names_gp = gpar(fontsize = 8.5, fontfamily = PLOT_FONT_FAMILY), rect_gp = gpar(type = "none"), border = TRUE, left_annotation = if (row_side == 
        "right") 
        anno
    else NULL, right_annotation = if (row_side == "left") 
        anno
    else NULL, cell_fun = function(j, i, x, y, w, h, fill) {
        val <- mat[i, j]
        mag <- min(abs(val)/lim, 1)
        rr_mm <- BUBBLE_RADIUS_MIN_MM + (BUBBLE_RADIUS_MAX_MM - BUBBLE_RADIUS_MIN_MM) * mag
        grid.circle(x, y, r = grid::unit(rr_mm, "mm"), gp = gpar(fill = col_fun(max(min(val, lim), -lim)), col = "#BDBDBD", lwd = 0.45))
        if (star[i, j] != "") 
            grid.text(star[i, j], x, y, gp = gpar(fontsize = 7, fontfamily = PLOT_FONT_FAMILY))
    })
    h <- max(BUBBLE_MIN_HEIGHT_IN, length(feats) * BUBBLE_ROW_HEIGHT_IN + BUBBLE_BASE_HEIGHT_IN)
    pdf_w <- pdf_width_in
    safe_cairo_pdf(outfile, width = pdf_w, height = h)
    ComplexHeatmap::draw(ht, padding = grid::unit(c(0.35, 0.35, 0.35, 0.35), "mm"))
    if (!is.null(axis_title) && nzchar(axis_title)) {
        grid::grid.text(axis_title, x = grid::unit(1.5, "mm"), y = grid::unit(0.5, "npc"), rot = 90, gp = grid::gpar(fontfamily = PLOT_FONT_FAMILY, fontsize = 8.5, fontface = "bold"))
    }
    dev.off()
    display_heatmap_inline(ht, width = pdf_w, height = h)
}
```

```r
make_feature_sankey_node_dict <- function(axis_i) {
    ko_tbl_i <- ko_stats %>% filter(axis == axis_i, in_main) %>% arrange(subpathway_en, main_rank) %>% distinct(ko_id, .keep_all = TRUE)
    met_tbl_i <- selected_mets %>% filter(axis == axis_i, in_main) %>% arrange(subpathway_en, main_rank) %>% distinct(metabolite, .keep_all = TRUE)
    if (nrow(ko_tbl_i) == 0 || nrow(met_tbl_i) == 0) {
        return(NULL)
    }
    src <- ko_tbl_i %>% mutate(node_id = paste0("KO::", ko_id), display_name = ko_alias_display, side = "source", module = subpathway_en, color = unname(ko_module_colors[subpathway_en]), plot_order = row_number()) %>% transmute(node_id, display_name, side, module, color, plot_order, raw_feature = ko_id)
    tgt <- met_tbl_i %>% mutate(node_id = paste0("MET::", metabolite), display_name = short_metabolite_label(metabolite, source), side = "target", module = subpathway_en, color = unname(met_module_colors[subpathway_en]), plot_order = row_number()) %>% transmute(node_id, display_name, side, module, color, plot_order, raw_feature = metabolite)
    bind_rows(src, tgt) %>% distinct(node_id, .keep_all = TRUE)
}
calc_feature_sankey_flow <- function(axis_i, samples, panel_label) {
    ko_tbl_i <- ko_stats %>% filter(axis == axis_i, in_main) %>% distinct(ko_id, .keep_all = TRUE)
    met_tbl_i <- selected_mets %>% filter(axis == axis_i, in_main) %>% distinct(metabolite, .keep_all = TRUE)
    kos <- intersect(ko_tbl_i$ko_id, rownames(kegg_in))
    mets <- intersect(met_tbl_i$metabolite, rownames(metab_in))
    ss <- Reduce(intersect, list(samples, colnames(kegg_in), colnames(metab_in)))
    if (length(kos) == 0 || length(mets) == 0 || length(ss) < MIN_GROUP_N) {
        return(NULL)
    }
    sig_pairs <- ko_met_edge_stats %>% filter(axis == axis_i, significant, anchor %in% kos, target %in% mets) %>% arrange(q_value, desc(abs(rho))) %>% group_by(anchor, target) %>% slice(1) %>% ungroup() %>% transmute(source_KO = anchor, metabolite = target, discovery_rho = rho, discovery_p = p_value, discovery_q = q_value, metabolite_subtype, fdr_family)
    if (nrow(sig_pairs) == 0) {
        return(NULL)
    }
    pair_stats <- lapply(seq_len(nrow(sig_pairs)), function(ii) {
        ko_i <- sig_pairs$source_KO[ii]
        met_i <- sig_pairs$metabolite[ii]
        x <- as.numeric(kegg_in[ko_i, ss, drop = TRUE])
        y <- as.numeric(metab_in[met_i, ss, drop = TRUE])
        ct <- safe_cor_test(x, y)
        data.frame(source_KO = ko_i, metabolite = met_i, rho = as.numeric(ct["rho"]), p_value = as.numeric(ct["p"]), n_pair = as.numeric(ct["n"]), stringsAsFactors = FALSE)
    }) %>% bind_rows()
    out <- sig_pairs %>% left_join(pair_stats, by = c("source_KO", "metabolite")) %>% left_join(ko_tbl_i %>% transmute(source_KO = ko_id, source_label = ko_alias_display, source_module = subpathway_en), by = "source_KO") %>% left_join(met_tbl_i %>% transmute(metabolite, metabolite_label = short_metabolite_label(metabolite, source), target_module = subpathway_en, metabolite_source = source), by = "metabolite") %>% mutate(source_node = paste0("KO::", source_KO), target_node = paste0("MET::", metabolite), 
        thickness = abs(rho), direction = case_when(!is.finite(rho) ~ "NA", rho >= 0 ~ "Positive", TRUE ~ "Negative"), panel = panel_label, axis = axis_i) %>% filter(is.finite(rho), is.finite(thickness), thickness > 0) %>% arrange(source_module, source_KO, target_module, metabolite)
    out
}
for (axis_i in AXIS_LEVELS) {
    safe_axis <- safe_name(axis_i)
    node_dict <- make_feature_sankey_node_dict(axis_i)
    if (is.null(node_dict) || nrow(node_dict) == 0) {
        message("[Feature Sankey skipped] ", axis_i)
        next
    }
    flow_nr <- calc_feature_sankey_flow(axis_i, response_groups$`Non-Responder`, "Non-Responder")
    flow_r <- calc_feature_sankey_flow(axis_i, response_groups$Responder, "Responder")
    active_response_nodes <- unique(c(if (!is.null(flow_nr)) c(flow_nr$source_node, flow_nr$target_node) else character(), if (!is.null(flow_r)) c(flow_r$source_node, flow_r$target_node) else character()))
    node_dict_response <- node_dict %>% filter(node_id %in% active_response_nodes)
    if (nrow(node_dict_response) > 0) {
        write.csv(node_dict_response, file.path(dirs$main_sankey, paste0(safe_axis, get_value("science.tables.result_265", "result_265.csv"))), row.names = FALSE)
    }
    if (!is.null(flow_nr) && nrow(flow_nr) > 0) {
        write.csv(flow_nr, file.path(dirs$main_sankey, paste0(safe_axis, get_value("science.tables.result_266", "result_266.csv"))), row.names = FALSE)
    }
    if (!is.null(flow_r) && nrow(flow_r) > 0) {
        write.csv(flow_r, file.path(dirs$main_sankey, paste0(safe_axis, get_value("science.tables.result_267", "result_267.csv"))), row.names = FALSE)
    }
    if (length(mbp_groups$`MBP-I`) >= MIN_GROUP_N && length(mbp_groups$`MBP-III`) >= MIN_GROUP_N) {
        flow_i <- calc_feature_sankey_flow(axis_i, mbp_groups$`MBP-I`, "MBP-I")
        flow_iii <- calc_feature_sankey_flow(axis_i, mbp_groups$`MBP-III`, "MBP-III")
        active_mbp_nodes <- unique(c(if (!is.null(flow_i)) c(flow_i$source_node, flow_i$target_node) else character(), if (!is.null(flow_iii)) c(flow_iii$source_node, flow_iii$target_node) else character()))
        node_dict_mbp <- node_dict %>% filter(node_id %in% active_mbp_nodes)
        if (nrow(node_dict_mbp) > 0) {
            write.csv(node_dict_mbp, file.path(dirs$mbp_sankey, paste0(safe_axis, get_value("science.tables.result_265", "result_265.csv"))), row.names = FALSE)
        }
        if (!is.null(flow_i) && nrow(flow_i) > 0) {
            write.csv(flow_i, file.path(dirs$mbp_sankey, paste0(safe_axis, get_value("science.tables.result_268", "result_268.csv"))), row.names = FALSE)
        }
        if (!is.null(flow_iii) && nrow(flow_iii) > 0) {
            write.csv(flow_iii, file.path(dirs$mbp_sankey, paste0(safe_axis, get_value("science.tables.result_269", "result_269.csv"))), row.names = FALSE)
        }
    }
}
```

```r
short_radiomics_label <- function(x) {
    z <- as.character(x)
    z <- str_replace(z, regex("^original_", ignore_case = TRUE), "Original | ")
    z <- str_replace(z, regex("^wavelet-([^_]+)_", ignore_case = TRUE), "Wavelet-\\1 | ")
    z <- str_replace(z, regex("^log-sigma-([^_]+)_", ignore_case = TRUE), "LoG-\\1 | ")
    z <- str_replace_all(z, "_", " ")
    compact_middle_label(z, max_chars = RADIOMICS_LABEL_MAX_CHARS)
}
select_display_metabolites <- function(axis_i) {
    available_mets <- selected_mets %>% filter(axis == axis_i) %>% distinct(metabolite, .keep_all = TRUE)
    if (nrow(available_mets) == 0) {
        return(data.frame())
    }
    sig_edge_i <- met_rad_edge_stats %>% filter(axis == axis_i, significant, anchor %in% available_mets$metabolite)
    if (nrow(sig_edge_i) > 0) {
        edge_supported_rank <- sig_edge_i %>% group_by(anchor) %>% summarise(n_sig_radiomics = n_distinct(target), min_rad_edge_q = suppressWarnings(min(q_value, na.rm = TRUE)), max_rad_abs_rho = suppressWarnings(max(abs(rho), na.rm = TRUE)), .groups = "drop") %>% mutate(min_rad_edge_q = ifelse(is.finite(min_rad_edge_q), min_rad_edge_q, NA_real_), max_rad_abs_rho = ifelse(is.finite(max_rad_abs_rho), max_rad_abs_rho, NA_real_)) %>% arrange(desc(n_sig_radiomics), min_rad_edge_q, desc(max_rad_abs_rho))
    }
    else {
        edge_supported_rank <- data.frame(anchor = character(), n_sig_radiomics = integer(), min_rad_edge_q = numeric(), max_rad_abs_rho = numeric(), stringsAsFactors = FALSE)
    }
    primary_mets <- head(edge_supported_rank$anchor, MAIN_RAD_MATCHED_METAB_N)
    n_need <- max(0, MIN_METAB_ROWS_PER_AXIS - length(primary_mets))
    if (n_need > 0) {
        filler_pool <- available_mets %>% filter(!metabolite %in% primary_mets) %>% arrange(desc(n_significant_links_max), desc(biomap_score_max), response_q_axis_source, response_p, min_edge_q, desc(max_abs_rho), main_rank)
        filler_mets <- head(filler_pool$metabolite, n_need)
    }
    else {
        filler_mets <- character()
    }
    final_mets <- unique(c(primary_mets, filler_mets))
    final_mets <- head(final_mets, max(MIN_METAB_ROWS_PER_AXIS, MAIN_RAD_MATCHED_METAB_N))
    available_mets %>% filter(metabolite %in% final_mets) %>% mutate(radiomics_edge_supported = metabolite %in% edge_supported_rank$anchor, display_role = ifelse(radiomics_edge_supported, "Radiomics-edge-supported", "KO-metabolite-supported filler"), display_order = match(metabolite, final_mets)) %>% arrange(display_order)
}
select_display_radiomics <- function(axis_i, display_mets) {
    if (nrow(display_mets) == 0) {
        return(data.frame())
    }
    met_names <- display_mets$metabolite
    available_rad <- selected_rad %>% filter(axis == axis_i) %>% distinct(radiomics_feature, .keep_all = TRUE)
    if (nrow(available_rad) == 0) {
        return(data.frame())
    }
    sig_edge_i <- met_rad_edge_stats %>% filter(axis == axis_i, significant, anchor %in% met_names, target %in% available_rad$radiomics_feature)
    rad_seed <- character()
    if (nrow(sig_edge_i) > 0) {
        rad_seed <- sig_edge_i %>% group_by(anchor) %>% arrange(q_value, desc(abs(rho)), .by_group = TRUE) %>% slice_head(n = RAD_SEED_PER_METAB) %>% ungroup() %>% pull(target) %>% unique()
    }
    if (nrow(sig_edge_i) > 0) {
        rad_network_rank <- sig_edge_i %>% group_by(target) %>% summarise(n_display_met_links = n_distinct(anchor), min_edge_q = suppressWarnings(min(q_value, na.rm = TRUE)), max_abs_rho = suppressWarnings(max(abs(rho), na.rm = TRUE)), .groups = "drop") %>% left_join(available_rad %>% select(radiomics_feature, response_q_category, response_p, radiomics_module, radiomics_filter, main_rank), by = c(target = "radiomics_feature")) %>% arrange(desc(n_display_met_links), response_q_category, min_edge_q, 
            desc(max_abs_rho), main_rank)
    }
    else {
        rad_network_rank <- data.frame(target = character(), stringsAsFactors = FALSE)
    }
    matched_rads <- unique(c(rad_seed, rad_network_rank$target))
    matched_rads <- matched_rads[!is.na(matched_rads)]
    n_need <- max(0, MAIN_RADIOMICS_DISPLAY - length(matched_rads))
    if (n_need > 0) {
        filler_rad <- available_rad %>% filter(!radiomics_feature %in% matched_rads) %>% arrange(response_q_category, response_p, desc(n_significant_links_max), desc(biomap_score_max), min_edge_q, desc(max_abs_rho), main_rank) %>% slice_head(n = n_need) %>% pull(radiomics_feature)
        matched_rads <- unique(c(matched_rads, filler_rad))
    }
    matched_rads <- head(matched_rads, MAIN_RADIOMICS_DISPLAY)
    available_rad %>% filter(radiomics_feature %in% matched_rads) %>% mutate(display_order = match(radiomics_feature, matched_rads)) %>% arrange(display_order)
}
```

```r
read_external_label_source <- function(candidates) {
    existing <- candidates[file.exists(candidates)]
    if (length(existing) == 0) {
        stop("No external label file found. Checked:\n", paste(candidates, collapse = "\n"))
    }
    path <- existing[1]
    if (grepl("\\.xlsx?$", path, ignore.case = TRUE)) {
        df <- suppressMessages(readxl::read_excel(path, sheet = 1)) %>% as.data.frame(check.names = FALSE)
    }
    else {
        df <- read_any_csv(path)
    }
    id_col <- detect_col(df, c("sample_id", "Sample ID", "sample", "sample_id", "patient", "patient_id", "ID", "id"), TRUE, "external patient/sample ID")
    response_col <- detect_col(df, c("Response", "response", "Group", "group", get_column("outcomes.response", "Response"), "mRECIST_response", "response_label"), TRUE, "external Response")
    labels <- data.frame(sample = standardize_id(df[[id_col]]), Response = normalize_response(df[[response_col]]), stringsAsFactors = FALSE) %>% filter(!is.na(sample), sample != "", !is.na(Response)) %>% distinct(sample, .keep_all = TRUE)
    list(path = path, labels = labels)
}
load_external_branch <- function(file, label_tbl, branch_name, expected_n = NA_integer_, suffix = NULL, normalize_ko = FALSE) {
    if (!file.exists(file)) {
        stop("External ", branch_name, " matrix not found: ", file)
    }
    mat <- read_matrix_auto(file, label_tbl$sample, "feature_by_sample")
    colnames(mat) <- standardize_id(colnames(mat))
    if (normalize_ko) {
        ko_std <- str_extract(toupper(rownames(mat)), "K\\d{5}")
        valid <- !is.na(ko_std)
        if (mean(valid) > 0.8 && anyDuplicated(ko_std[valid]) == 0) {
            rownames(mat)[valid] <- ko_std[valid]
        }
    }
    if (!is.null(suffix)) {
        mat <- add_suffix_if_missing(mat, suffix)
    }
    matrix_samples <- colnames(mat)
    common <- intersect(matrix_samples, label_tbl$sample)
    meta_i <- label_tbl %>% filter(sample %in% common)
    groups_i <- list(Responder = meta_i$sample[meta_i$Response == "Responder"], `Non-Responder` = meta_i$sample[meta_i$Response == "Non-Responder"])
    if (length(groups_i$Responder) < MIN_GROUP_N || length(groups_i$`Non-Responder`) < MIN_GROUP_N) {
        stop("External ", branch_name, " has too few labeled samples in one Response group: R=", length(groups_i$Responder), ", NR=", length(groups_i$`Non-Responder`))
    }
    if (is.finite(expected_n) && ncol(mat) != expected_n) {
        warning("External ", branch_name, " matrix sample count = ", ncol(mat), "; expected audit target = ", expected_n, ". Analysis will use the actual matrix-label overlap.")
    }
    list(name = branch_name, file = file, matrix = mat, meta = meta_i, groups = groups_i, matrix_n = ncol(mat), labeled_n = length(common), responder_n = length(groups_i$Responder), nonresponder_n = length(groups_i$`Non-Responder`))
}
calc_fixed_feature_effect_stats <- function(mat, features, group_r, group_nr) {
    features <- unique(features[features %in% rownames(mat)])
    if (length(features) == 0) {
        return(data.frame())
    }
    st <- feature_group_effect(mat[features, , drop = FALSE], group_r, group_nr) %>% mutate(q_BH = bh_adjust_safe(p), star_nominal = p_to_star(p), star_fdr = p_to_star(q_BH))
    st
}
choose_external_star <- function(stat_df) {
    if (toupper(EXTERNAL_STAR_MODE) == "FDR") {
        stat_df$star_fdr
    }
    else {
        stat_df$star_nominal
    }
}
plot_internal_external_bubble <- function(feature_tbl, feature_col, label_col, module_col, module_colors, internal_stats, external_stats, outfile, external_column_label, row_side = "right", body_col_width_mm = BUBBLE_COL_WIDTH_MM, row_label_max_width_mm = ROW_LABEL_MAX_WIDTH_MM, annotation_width_mm = 3, pdf_width_in = BUBBLE_DOUBLE_PDF_WIDTH_IN, show_module_titles = FALSE, show_column_names = TRUE, column_names_rot = BUBBLE_COLUMN_LABEL_ROT) {
    feats <- unique(feature_tbl[[feature_col]])
    feats <- feats[feats %in% internal_stats$feature & feats %in% external_stats$feature]
    if (length(feats) == 0) {
        return(NULL)
    }
    ft <- feature_tbl[match(feats, feature_tbl[[feature_col]]), , drop = FALSE]
    st_in <- internal_stats[match(feats, internal_stats$feature), , drop = FALSE]
    st_ex <- external_stats[match(feats, external_stats$feature), , drop = FALSE]
    mat <- cbind(Internal = st_in$effect, External = st_ex$effect)
    colnames(mat)[2] <- external_column_label
    stars <- cbind(Internal = choose_external_star(st_in), External = choose_external_star(st_ex))
    colnames(stars)[2] <- external_column_label
    rownames(mat) <- feats
    rownames(stars) <- feats
    lim <- BUBBLE_EFFECT_LIM
    col_fun <- circlize::colorRamp2(c(-lim, 0, lim), c("#3C5488", "white", "#E64B35"))
    module_vec <- as.character(ft[[module_col]])
    module_vec[is.na(module_vec) | module_vec == ""] <- "Unclassified"
    module_display_vec <- if (isTRUE(show_module_titles)) {
        ko_module_display_label(module_vec)
    }
    else {
        module_vec
    }
    module_factor <- factor(module_display_vec, levels = unique(module_display_vec))
    anno <- rowAnnotation(Module = anno_simple(module_vec, col = module_colors, width = unit(annotation_width_mm, "mm")), show_annotation_name = FALSE, show_legend = FALSE)
    labels <- as.character(ft[[label_col]])
    bad_lab <- is.na(labels) | labels == ""
    labels[bad_lab] <- feats[bad_lab]
    ht <- Heatmap(mat, name = "Effect", col = col_fun, show_heatmap_legend = FALSE, width = grid::unit(ncol(mat) * body_col_width_mm, "mm"), cluster_rows = FALSE, cluster_columns = FALSE, row_split = module_factor, row_title = if (isTRUE(show_module_titles)) 
        "%s"
    else NULL, row_title_rot = 0, row_title_side = "left", row_title_gp = gpar(fontsize = if (isTRUE(show_module_titles)) 
        KO_MODULE_TITLE_FONT_PT
    else 0, fontfamily = PLOT_FONT_FAMILY, fontface = "bold"), row_gap = grid::unit(0.65, "mm"), show_row_names = TRUE, row_labels = labels, row_names_side = row_side, row_names_gp = gpar(fontsize = 7.8, fontfamily = PLOT_FONT_FAMILY), row_names_max_width = grid::unit(row_label_max_width_mm, "mm"), show_column_names = show_column_names, column_names_rot = column_names_rot, column_names_gp = gpar(fontsize = 8, fontfamily = PLOT_FONT_FAMILY), rect_gp = gpar(type = "none"), border = TRUE, left_annotation = if (row_side == 
        "right") {
        anno
    }
    else {
        NULL
    }, right_annotation = if (row_side == "left") {
        anno
    }
    else {
        NULL
    }, cell_fun = function(j, i, x, y, w, h, fill) {
        val <- mat[i, j]
        mag <- min(abs(val)/lim, 1)
        rr_mm <- BUBBLE_RADIUS_MIN_MM + (BUBBLE_RADIUS_MAX_MM - BUBBLE_RADIUS_MIN_MM) * mag
        grid.circle(x, y, r = grid::unit(rr_mm, "mm"), gp = gpar(fill = col_fun(max(min(val, lim), -lim)), col = "#BDBDBD", lwd = 0.45))
        if (!is.na(stars[i, j]) && stars[i, j] != "") {
            grid.text(stars[i, j], x, y, gp = gpar(fontsize = 7, fontfamily = PLOT_FONT_FAMILY))
        }
    })
    h <- max(BUBBLE_MIN_HEIGHT_IN, length(feats) * BUBBLE_ROW_HEIGHT_IN + BUBBLE_BASE_HEIGHT_IN)
    pdf_w <- pdf_width_in
    safe_cairo_pdf(outfile, width = pdf_w, height = h)
    ComplexHeatmap::draw(ht, padding = grid::unit(c(0.35, 0.35, 0.35, 0.35), "mm"))
    dev.off()
    display_heatmap_inline(ht, width = pdf_w, height = h)
}
if (RUN_EXTERNAL_VALIDATION) {
    ext_label_source <- read_external_label_source(EXTERNAL_LABEL_CANDIDATES)
    ext_labels <- ext_label_source$labels
    ext_ko <- load_external_branch(KO_EXTERNAL_FILE, ext_labels, "KO", EXPECTED_EXTERNAL_N["KO"], suffix = NULL, normalize_ko = TRUE)
    ext_blood <- load_external_branch(BLOOD_EXTERNAL_FILE, ext_labels, "Blood", EXPECTED_EXTERNAL_N["Blood"], suffix = "(B)", normalize_ko = FALSE)
    ext_urine <- load_external_branch(URINE_EXTERNAL_FILE, ext_labels, "Urine", EXPECTED_EXTERNAL_N["Urine"], suffix = "(U)", normalize_ko = FALSE)
    ext_radiomics <- load_external_branch(RADIOMICS_EXTERNAL_FILE, ext_labels, "Radiomics", EXPECTED_EXTERNAL_N["Radiomics"], suffix = NULL, normalize_ko = FALSE)
    external_cohort_audit <- bind_rows(data.frame(branch = ext_ko$name, matrix_file = ext_ko$file, matrix_n = ext_ko$matrix_n, labeled_overlap_n = ext_ko$labeled_n, responder_n = ext_ko$responder_n, nonresponder_n = ext_ko$nonresponder_n), data.frame(branch = ext_blood$name, matrix_file = ext_blood$file, matrix_n = ext_blood$matrix_n, labeled_overlap_n = ext_blood$labeled_n, responder_n = ext_blood$responder_n, nonresponder_n = ext_blood$nonresponder_n), data.frame(branch = ext_urine$name, matrix_file = ext_urine$file, 
        matrix_n = ext_urine$matrix_n, labeled_overlap_n = ext_urine$labeled_n, responder_n = ext_urine$responder_n, nonresponder_n = ext_urine$nonresponder_n), data.frame(branch = ext_radiomics$name, matrix_file = ext_radiomics$file, matrix_n = ext_radiomics$matrix_n, labeled_overlap_n = ext_radiomics$labeled_n, responder_n = ext_radiomics$responder_n, nonresponder_n = ext_radiomics$nonresponder_n))
    external_cohort_audit$label_source <- ext_label_source$path
    write.csv(external_cohort_audit, file.path(dirs$external_tables, get_value("science.tables.result_270", "result_270.csv")), row.names = FALSE)
    external_stats_list <- list()
    ee <- 1
    for (axis_i in AXIS_LEVELS) {
        safe_axis <- safe_name(axis_i)
        show_external_bottom_labels <- if (isTRUE(EXTERNAL_ONLY_BOTTOM_COLUMN_LABELS)) {
            identical(axis_i, utils::tail(AXIS_LEVELS, 1))
        }
        else {
            TRUE
        }
        ko_tbl_i <- ko_stats %>% filter(axis == axis_i, in_main) %>% arrange(main_rank)
        if (nrow(ko_tbl_i) > 0) {
            ko_feats <- ko_tbl_i$ko_id
            st_ko_internal <- calc_fixed_feature_effect_stats(kegg_in, ko_feats, response_groups$Responder, response_groups$`Non-Responder`)
            st_ko_external <- calc_fixed_feature_effect_stats(ext_ko$matrix, ko_feats, ext_ko$groups$Responder, ext_ko$groups$`Non-Responder`)
            ko_stats_export <- full_join(st_ko_internal %>% transmute(feature, internal_effect = effect, internal_p = p, internal_q_display = q_BH), st_ko_external %>% transmute(feature, external_effect = effect, external_p = p, external_q_display = q_BH), by = "feature") %>% mutate(axis = axis_i, modality = "KO", direction_concordant = sign(internal_effect) == sign(external_effect))
            external_stats_list[[ee]] <- ko_stats_export
            ee <- ee + 1
            plot_internal_external_bubble(ko_tbl_i, "ko_id", "ko_alias_display", "subpathway_en", ko_module_colors, st_ko_internal, st_ko_external, file.path(dirs$external_ko, paste0(safe_axis, get_value("science.figures.plot_106", "plot_106.pdf"))), "External", row_side = "right", body_col_width_mm = KO_BUBBLE_COL_WIDTH_MM, row_label_max_width_mm = KO_ROW_LABEL_MAX_WIDTH_MM, annotation_width_mm = KO_ANNOTATION_WIDTH_MM, pdf_width_in = KO_DOUBLE_PDF_WIDTH_IN, show_module_titles = KO_SHOW_MODULE_TITLES, 
                show_column_names = show_external_bottom_labels, column_names_rot = BUBBLE_COLUMN_LABEL_ROT)
        }
        met_tbl_i <- selected_mets %>% filter(axis == axis_i, in_main) %>% arrange(main_rank) %>% mutate(display_label = short_metabolite_label(metabolite, source))
        if (nrow(met_tbl_i) > 0) {
            met_feats <- met_tbl_i$metabolite
            st_met_internal <- calc_fixed_feature_effect_stats(metab_in, met_feats, response_groups$Responder, response_groups$`Non-Responder`)
            blood_feats <- met_tbl_i$metabolite[met_tbl_i$source == "Blood"]
            urine_feats <- met_tbl_i$metabolite[met_tbl_i$source == "Urine"]
            st_blood_external <- calc_fixed_feature_effect_stats(ext_blood$matrix, blood_feats, ext_blood$groups$Responder, ext_blood$groups$`Non-Responder`)
            st_urine_external <- calc_fixed_feature_effect_stats(ext_urine$matrix, urine_feats, ext_urine$groups$Responder, ext_urine$groups$`Non-Responder`)
            st_met_external <- bind_rows(st_blood_external, st_urine_external)
            met_stats_export <- full_join(st_met_internal %>% transmute(feature, internal_effect = effect, internal_p = p, internal_q_display = q_BH), st_met_external %>% transmute(feature, external_effect = effect, external_p = p, external_q_display = q_BH), by = "feature") %>% left_join(met_tbl_i %>% select(metabolite, source) %>% distinct(metabolite, .keep_all = TRUE), by = c(feature = "metabolite")) %>% mutate(axis = axis_i, modality = source, direction_concordant = sign(internal_effect) == sign(external_effect))
            external_stats_list[[ee]] <- met_stats_export
            ee <- ee + 1
            plot_internal_external_bubble(met_tbl_i, "metabolite", "display_label", "subpathway_en", met_module_colors, st_met_internal, st_met_external, file.path(dirs$external_metab, paste0(safe_axis, get_value("science.figures.plot_107", "plot_107.pdf"))), "External", row_side = "left", body_col_width_mm = METAB_BUBBLE_COL_WIDTH_MM, row_label_max_width_mm = METAB_ROW_LABEL_MAX_WIDTH_MM, annotation_width_mm = METAB_ANNOTATION_WIDTH_MM, pdf_width_in = METAB_DOUBLE_PDF_WIDTH_IN + 1.55, show_module_titles = METAB_SHOW_MODULE_TITLES_EXTERNAL, 
                show_column_names = show_external_bottom_labels, column_names_rot = BUBBLE_COLUMN_LABEL_ROT)
        }
    }
    external_feature_stats <- bind_rows(external_stats_list)
    write.csv(external_feature_stats, file.path(dirs$external_tables, get_value("science.tables.result_271", "result_271.csv")), row.names = FALSE)
    external_concordance_summary <- external_feature_stats %>% filter(is.finite(internal_effect), is.finite(external_effect)) %>% group_by(axis, modality) %>% summarise(n_features = n(), n_direction_concordant = sum(direction_concordant, na.rm = TRUE), direction_concordance_rate = mean(direction_concordant, na.rm = TRUE), effect_spearman_rho = if (n() >= 3) {
        suppressWarnings(cor(internal_effect, external_effect, method = "spearman", use = "complete.obs"))
    }
    else {
        NA_real_
    }, .groups = "drop")
    write.csv(external_concordance_summary, file.path(dirs$external_tables, get_value("science.tables.metrics_085", "metrics_085.csv")), row.names = FALSE)
}
```

```r
standardize_id <- function(x) {
    toupper(trimws(as.character(x)))
}
read_any_csv <- function(file) {
    if (!file.exists(file)) {
        stop("Missing file: ", file)
    }
    as.data.frame(data.table::fread(file, data.table = FALSE, check.names = FALSE), check.names = FALSE)
}
set_first_col_as_rownames <- function(x) {
    if (ncol(x) < 2) {
        return(x)
    }
    v <- x[[1]]
    nm <- colnames(x)[1]
    looks_id <- length(unique(v)) == nrow(x) && (is.character(v) || is.factor(v) || tolower(nm) %in% c("sample", "sample_id", "patient", "patient_id", "id", "feature", "name"))
    if (looks_id) {
        rownames(x) <- as.character(v)
        x[[1]] <- NULL
    }
    x
}
read_matrix_auto <- function(file, sample_ids, want = c("feature_by_sample", "sample_by_feature")) {
    want <- match.arg(want)
    x <- read_any_csv(file)
    sid <- standardize_id(sample_ids)
    if (ncol(x) >= 2) {
        first_ids <- standardize_id(x[[1]])
        if (sum(first_ids %in% sid, na.rm = TRUE) >= 2 && length(unique(first_ids)) == nrow(x)) {
            rownames(x) <- first_ids
            x[[1]] <- NULL
        }
        else {
            x <- set_first_col_as_rownames(x)
        }
    }
    rn_up <- standardize_id(rownames(x))
    cn_up <- standardize_id(colnames(x))
    row_hit <- sum(rn_up %in% sid)
    col_hit <- sum(cn_up %in% sid)
    if (want == "feature_by_sample") {
        if (row_hit > col_hit) {
            rownames(x) <- rn_up
            x <- as.data.frame(t(x), check.names = FALSE)
        }
        else {
            colnames(x) <- cn_up
        }
    }
    else {
        if (col_hit > row_hit) {
            colnames(x) <- cn_up
            x <- as.data.frame(t(x), check.names = FALSE)
        }
        else {
            rownames(x) <- rn_up
        }
    }
    x[] <- lapply(x, function(v) suppressWarnings(as.numeric(v)))
    x <- as.data.frame(x, check.names = FALSE)
    if (want == "feature_by_sample") {
        colnames(x) <- standardize_id(colnames(x))
    }
    else {
        rownames(x) <- standardize_id(rownames(x))
    }
    x
}
detect_col <- function(df, candidates, required = TRUE, label = "column") {
    nms_low <- tolower(colnames(df))
    idx <- match(tolower(candidates), nms_low)
    idx <- idx[!is.na(idx)]
    if (length(idx) > 0) {
        return(colnames(df)[idx[1]])
    }
    if (required) {
        stop("Cannot detect ", label, ". Available columns: ", paste(colnames(df), collapse = ", "))
    }
    NULL
}
normalize_response <- function(x) {
    xx <- str_to_lower(str_trim(as.character(x)))
    out <- rep(NA_character_, length(xx))
    out[xx %in% c("1", "r", "responder", "response", "cr", "pr", "yes", "true")] <- "Responder"
    out[xx %in% c("0", "nr", "non-responder", "nonresponder", "non responder", "sd", "pd", "no", "false")] <- "Non-Responder"
    out
}
normalize_mbp <- function(x) {
    xx <- toupper(str_trim(as.character(x)))
    out <- rep(NA_character_, length(xx))
    out[str_detect(xx, "^MBP[-_ ]?III$") | xx == "III" | xx == "3"] <- "MBP-III"
    out[str_detect(xx, "^MBP[-_ ]?II$") | xx == "II" | xx == "2"] <- "MBP-II"
    out[str_detect(xx, "^MBP[-_ ]?I$") | xx == "I" | xx == "1"] <- "MBP-I"
    out
}
safe_wilcox_p <- function(x, y) {
    x <- x[is.finite(x)]
    y <- y[is.finite(y)]
    if (length(x) < 3 || length(y) < 3) {
        return(NA_real_)
    }
    tryCatch(suppressWarnings(wilcox.test(x, y, exact = FALSE)$p.value), error = function(e) NA_real_)
}
safe_spearman <- function(x, y) {
    ok <- is.finite(x) & is.finite(y)
    if (sum(ok) < 5) {
        return(c(rho = NA_real_, p = NA_real_, n = sum(ok)))
    }
    z <- tryCatch(suppressWarnings(cor.test(x[ok], y[ok], method = "spearman", exact = FALSE)), error = function(e) NULL)
    if (is.null(z)) {
        return(c(rho = NA_real_, p = NA_real_, n = sum(ok)))
    }
    c(rho = unname(z$estimate), p = z$p.value, n = sum(ok))
}
safe_z <- function(x) {
    x <- as.numeric(x)
    s <- sd(x, na.rm = TRUE)
    if (sum(is.finite(x)) < 3 || !is.finite(s) || s == 0) {
        return(rep(NA_real_, length(x)))
    }
    as.numeric(scale(x))
}
safe_row_mean <- function(x, min_fraction = 0.7) {
    x <- as.matrix(x)
    if (ncol(x) == 0) {
        return(rep(NA_real_, nrow(x)))
    }
    n_ok <- rowSums(is.finite(x))
    min_n <- max(1L, ceiling(ncol(x) * min_fraction))
    out <- rowMeans(x, na.rm = TRUE)
    out[n_ok < min_n] <- NA_real_
    out[!is.finite(out)] <- NA_real_
    out
}
effect_z_two_group <- function(score, group, level_a = "Responder", level_b = "Non-Responder") {
    z <- safe_z(score)
    xa <- z[group == level_a]
    xb <- z[group == level_b]
    mean(xa, na.rm = TRUE) - mean(xb, na.rm = TRUE)
}
p_to_star <- function(p) {
    ifelse(is.na(p), "", ifelse(p < 0.001, "***", ifelse(p < 0.01, "**", ifelse(p < 0.05, "*", ""))))
}
```

```r
metabolic_module_members <- selected_mets %>% transmute(axis = as.character(axis), metabolic_module = as.character(metabolic_module_plot), metabolite = as.character(metabolite), source = ifelse(is.na(source) | source == "", "Unknown", as.character(source)), in_main = ifelse(is.na(in_main), FALSE, as.logical(in_main)), main_rank = suppressWarnings(as.numeric(main_rank))) %>% filter(!is.na(axis), axis != "", !is.na(metabolic_module), metabolic_module != "", !is.na(metabolite), metabolite != "") %>% 
    distinct(axis, metabolic_module, metabolite, .keep_all = TRUE) %>% mutate(module_id = paste(axis, metabolic_module, sep = " | "))
safe_col_mean <- function(x) {
    z <- colMeans(x, na.rm = TRUE)
    z[!is.finite(z)] <- NA_real_
    z
}
safe_row_mean <- function(x) {
    z <- rowMeans(x, na.rm = TRUE)
    z[!is.finite(z)] <- NA_real_
    z
}
score_metabolic_module <- function(member_df) {
    axis_i <- unique(member_df$axis)
    module_i <- unique(member_df$metabolic_module)
    module_id_i <- unique(member_df$module_id)
    if (length(axis_i) != 1 || length(module_i) != 1 || length(module_id_i) != 1) {
        stop("Metabolic module grouping is not unique.")
    }
    feats <- intersect(unique(member_df$metabolite), rownames(metab_internal))
    if (length(feats) == 0) 
        return(NULL)
    samples <- colnames(metab_internal)
    z_mat <- matrix(NA_real_, nrow = length(feats), ncol = length(samples), dimnames = list(feats, samples))
    for (f in feats) {
        z_mat[f, ] <- safe_z(as.numeric(metab_internal[f, samples, drop = TRUE]))
    }
    feat_source <- member_df$source[match(feats, member_df$metabolite)]
    feat_source[is.na(feat_source) | feat_source == ""] <- "Unknown"
    src_levels <- unique(feat_source)
    src_score <- matrix(NA_real_, nrow = length(samples), ncol = length(src_levels), dimnames = list(samples, src_levels))
    for (src_i in src_levels) {
        idx <- which(feat_source == src_i)
        src_score[, src_i] <- safe_col_mean(z_mat[idx, , drop = FALSE])
    }
    module_score <- safe_row_mean(src_score)
    main_rank_valid <- member_df$main_rank[member_df$in_main & is.finite(member_df$main_rank)]
    scores <- data.frame(sample = samples, axis = axis_i, metabolic_module = module_i, module_id = module_id_i, metabolic_module_score = as.numeric(module_score), n_metabolites = length(feats), n_sources = length(src_levels), main_display = any(member_df$in_main, na.rm = TRUE), min_main_rank = if (length(main_rank_valid) > 0) 
        min(main_rank_valid)
    else Inf, stringsAsFactors = FALSE)
    feature_qc <- bind_rows(lapply(feats, function(f) {
        ct <- safe_spearman(z_mat[f, ], module_score)
        data.frame(axis = axis_i, metabolic_module = module_i, module_id = module_id_i, metabolite = f, source = feat_source[match(f, feats)], rho_feature_to_module = as.numeric(ct["rho"]), p_feature_to_module = as.numeric(ct["p"]), n = as.numeric(ct["n"]), stringsAsFactors = FALSE)
    }))
    if (nrow(z_mat) >= 2) {
        cm <- suppressWarnings(cor(t(z_mat), method = "spearman", use = "pairwise.complete.obs"))
        pair_rho <- cm[upper.tri(cm)]
        pair_rho <- pair_rho[is.finite(pair_rho)]
    }
    else {
        pair_rho <- numeric()
    }
    feature_rho <- feature_qc$rho_feature_to_module
    feature_rho <- feature_rho[is.finite(feature_rho)]
    module_qc <- data.frame(axis = axis_i, metabolic_module = module_i, module_id = module_id_i, n_metabolites = length(feats), n_sources = length(src_levels), main_display = any(member_df$in_main, na.rm = TRUE), median_feature_to_module_rho = if (length(feature_rho) > 0) 
        median(feature_rho)
    else NA_real_, fraction_feature_to_module_positive = if (length(feature_rho) > 0) 
        mean(feature_rho > 0)
    else NA_real_, median_pairwise_metabolite_rho = if (length(pair_rho) > 0) 
        median(pair_rho)
    else NA_real_, singleton_module = length(feats) == 1, stringsAsFactors = FALSE)
    list(scores = scores, feature_qc = feature_qc, module_qc = module_qc)
}
module_split <- split(metabolic_module_members, metabolic_module_members$module_id)
module_obj <- lapply(module_split, score_metabolic_module)
module_obj <- module_obj[!vapply(module_obj, is.null, logical(1))]
metabolic_module_scores_long <- bind_rows(lapply(module_obj, function(x) x$scores))
metabolic_module_feature_qc <- bind_rows(lapply(module_obj, function(x) x$feature_qc))
metabolic_module_qc <- bind_rows(lapply(module_obj, function(x) x$module_qc))
write.csv(metabolic_module_scores_long, file.path(DIRS$tables, get_value("science.tables.result_287", "result_287.csv")), row.names = FALSE)
write.csv(metabolic_module_feature_qc, file.path(DIRS$tables, get_value("science.tables.result_288", "result_288.csv")), row.names = FALSE)
write.csv(metabolic_module_qc, file.path(DIRS$tables, get_value("science.tables.result_289", "result_289.csv")), row.names = FALSE)
imaging_module_wide <- internal_module_scores %>% transmute(sample = as.character(sample), imaging_module = as.character(module), imaging_score = as.numeric(module_score)) %>% pivot_wider(names_from = imaging_module, values_from = imaging_score)
metabolic_module_meta <- metabolic_module_scores_long %>% distinct(axis, metabolic_module, module_id, n_metabolites, n_sources, main_display, min_main_rank)
module_imaging_rows <- list()
kk <- 1L
for (i in seq_len(nrow(metabolic_module_meta))) {
    id_i <- metabolic_module_meta$module_id[i]
    met_i <- metabolic_module_scores_long %>% filter(module_id == id_i) %>% select(sample, metabolic_module_score)
    for (im_i in MODULE_LEVELS) {
        if (!im_i %in% colnames(imaging_module_wide)) 
            next
        img_i <- data.frame(sample = imaging_module_wide$sample, imaging_score = as.numeric(imaging_module_wide[, im_i, drop = TRUE]), stringsAsFactors = FALSE)
        dd <- inner_join(met_i, img_i, by = "sample")
        ct <- safe_spearman(dd$metabolic_module_score, dd$imaging_score)
        module_imaging_rows[[kk]] <- data.frame(axis = metabolic_module_meta$axis[i], metabolic_module = metabolic_module_meta$metabolic_module[i], module_id = id_i, n_metabolites = metabolic_module_meta$n_metabolites[i], n_sources = metabolic_module_meta$n_sources[i], main_display = metabolic_module_meta$main_display[i], min_main_rank = metabolic_module_meta$min_main_rank[i], imaging_module = im_i, rho = as.numeric(ct["rho"]), p = as.numeric(ct["p"]), n = as.numeric(ct["n"]), stringsAsFactors = FALSE)
        kk <- kk + 1L
    }
}
metabolic_module_imaging_stats <- bind_rows(module_imaging_rows) %>% group_by(axis) %>% mutate(q_BH_within_axis = p.adjust(p, method = "BH")) %>% ungroup() %>% mutate(q_lt_0_10 = is.finite(q_BH_within_axis) & q_BH_within_axis < 0.1)
write.csv(metabolic_module_imaging_stats, file.path(DIRS$tables, get_value("science.tables.result_290", "result_290.csv")), row.names = FALSE)
```

```r
SHAP_BRIDGE_DIR <- file.path(ROOT, get_value("science.paths.directory_163", "directory_163"), get_value("science.paths.directory_164", "directory_164"), get_value("science.paths.directory_165", "directory_165"))
dir.create(SHAP_BRIDGE_DIR, recursive = TRUE, showWarnings = FALSE)
required_bridge_objects <- c("AXIS_LEVELS", "ko_stats", "kegg_in", "metabolic_module_scores_long", "internal_module_scores", "MODULE_LEVELS", "ko_module_colors", "met_module_colors")
missing_bridge_objects <- required_bridge_objects[!vapply(required_bridge_objects, exists, logical(1), inherits = TRUE)]
if (length(missing_bridge_objects) > 0) {
    stop("SHAP bridge missing required object(s): ", paste(missing_bridge_objects, collapse = ", "), ". Run preceding Figure-6 sections first.")
}
bridge_z <- function(x) {
    x <- as.numeric(x)
    mu <- mean(x, na.rm = TRUE)
    ss <- stats::sd(x, na.rm = TRUE)
    if (!is.finite(mu) || !is.finite(ss) || ss <= 0) {
        return(rep(NA_real_, length(x)))
    }
    (x - mu)/ss
}
bridge_col_mean <- function(x) {
    z <- colMeans(as.matrix(x), na.rm = TRUE)
    z[!is.finite(z)] <- NA_real_
    z
}
bridge_named_color <- function(key, palette, default = "#BDBDBD") {
    z <- unname(palette[as.character(key)])
    if (length(z) == 0 || is.na(z[1]) || !nzchar(z[1])) 
        return(default)
    z[1]
}
bridge_module_id <- function(prefix, axis, module) {
    paste(prefix, axis, module, sep = "::")
}
ko_members_shap <- ko_stats %>% filter(in_main) %>% transmute(axis = as.character(axis), ko_module = as.character(subpathway_en), ko_id = as.character(ko_id), main_rank = suppressWarnings(as.numeric(main_rank))) %>% filter(!is.na(axis), axis != "", !is.na(ko_module), ko_module != "", !is.na(ko_id), ko_id != "") %>% distinct(axis, ko_module, ko_id, .keep_all = TRUE) %>% mutate(module_id = bridge_module_id("KO", axis, ko_module))
ko_score_list_shap <- lapply(split(ko_members_shap, ko_members_shap$module_id), function(member_df) {
    axis_i <- unique(member_df$axis)
    module_i <- unique(member_df$ko_module)
    module_id_i <- unique(member_df$module_id)
    if (length(axis_i) != 1 || length(module_i) != 1 || length(module_id_i) != 1) {
        stop("KO module definition is not unique in SHAP bridge.")
    }
    feats <- intersect(unique(member_df$ko_id), rownames(kegg_in))
    if (length(feats) == 0) 
        return(NULL)
    samples <- colnames(kegg_in)
    z_mat <- matrix(NA_real_, nrow = length(feats), ncol = length(samples), dimnames = list(feats, samples))
    for (f in feats) {
        z_mat[f, ] <- bridge_z(kegg_in[f, samples, drop = TRUE])
    }
    data.frame(sample = standardize_id(samples), axis = axis_i, ko_module = module_i, module_id = module_id_i, module_score = as.numeric(bridge_col_mean(z_mat)), n_features = length(feats), stringsAsFactors = FALSE)
})
ko_score_long_shap <- bind_rows(ko_score_list_shap)
if (nrow(ko_score_long_shap) == 0) stop("No KO module scores could be generated.")
ko_meta_shap <- ko_members_shap %>% group_by(axis, ko_module, module_id) %>% summarise(n_features = n_distinct(ko_id), local_order = {
    z <- main_rank[is.finite(main_rank)]
    if (length(z) > 0) 
        min(z)
    else 999
}, .groups = "drop") %>% mutate(display_label = ko_module, module_type = "Microbial KO module", row_color = vapply(ko_module, bridge_named_color, character(1), palette = ko_module_colors), axis_index = match(axis, AXIS_LEVELS), row_order = axis_index * 100 + local_order)
met_score_long_shap <- metabolic_module_scores_long %>% mutate(sample = standardize_id(sample), axis = as.character(axis), metabolic_module = as.character(metabolic_module), main_display = as.logical(main_display), module_id_shap = bridge_module_id("Metab", axis, metabolic_module)) %>% filter(main_display) %>% transmute(sample, axis, metabolic_module, module_id = module_id_shap, module_score = as.numeric(metabolic_module_score), n_features = suppressWarnings(as.numeric(n_metabolites)), n_sources = suppressWarnings(as.numeric(n_sources)), 
    min_main_rank = suppressWarnings(as.numeric(min_main_rank)))
if (nrow(met_score_long_shap) == 0) {
    stop("No main-display metabolic module scores were available.")
}
met_meta_shap <- metabolic_module_scores_long %>% mutate(axis = as.character(axis), metabolic_module = as.character(metabolic_module), main_display = as.logical(main_display), module_id = bridge_module_id("Metab", axis, metabolic_module)) %>% filter(main_display) %>% group_by(axis, metabolic_module, module_id) %>% summarise(n_features = max(suppressWarnings(as.numeric(n_metabolites)), na.rm = TRUE), n_sources = max(suppressWarnings(as.numeric(n_sources)), na.rm = TRUE), local_order = {
    z <- suppressWarnings(as.numeric(min_main_rank))
    z <- z[is.finite(z)]
    if (length(z) > 0) 
        min(z)
    else 999
}, .groups = "drop") %>% mutate(display_label = metabolic_module, module_type = "Metabolite module", row_color = vapply(metabolic_module, bridge_named_color, character(1), palette = met_module_colors), axis_index = match(axis, AXIS_LEVELS), row_order = axis_index * 100 + 50 + local_order)
imaging_color_shap <- c(`Low-gray-weighted phenotype` = "#5B9BD5", `Coarse structural organization` = "#7F95B8", `Fine-scale heterogeneity` = "#6F6FAE")
img_score_long_shap <- internal_module_scores %>% transmute(sample = standardize_id(sample), imaging_module = as.character(module), module_score = as.numeric(module_score)) %>% filter(imaging_module %in% MODULE_LEVELS) %>% mutate(module_id = paste0("Rad::", imaging_module)) %>% select(sample, imaging_module, module_id, module_score)
if (nrow(img_score_long_shap) == 0) {
    stop("No final imaging phenotype scores were available.")
}
img_meta_shap <- data.frame(axis = rep("Radiomics", length(MODULE_LEVELS)), imaging_module = MODULE_LEVELS, module_id = paste0("Rad::", MODULE_LEVELS), display_label = MODULE_LEVELS, module_type = rep("Radiomics module", length(MODULE_LEVELS)), row_color = unname(imaging_color_shap[MODULE_LEVELS]), row_order = 10000 + seq_along(MODULE_LEVELS), n_features = NA_real_, stringsAsFactors = FALSE)
common_shap_samples <- Reduce(intersect, list(unique(ko_score_long_shap$sample), unique(met_score_long_shap$sample), unique(img_score_long_shap$sample)))
common_shap_samples <- sort(unique(common_shap_samples))
if (length(common_shap_samples) == 0) {
    stop("No common internal patients across KO, metabolic and imaging module scores.")
}
shap_module_scores_long <- bind_rows(ko_score_long_shap %>% filter(sample %in% common_shap_samples) %>% transmute(sample, module_id, module_score), met_score_long_shap %>% filter(sample %in% common_shap_samples) %>% transmute(sample, module_id, module_score), img_score_long_shap %>% filter(sample %in% common_shap_samples) %>% transmute(sample, module_id, module_score))
duplicate_module_scores <- shap_module_scores_long %>% count(sample, module_id, name = "n") %>% filter(n > 1)
if (nrow(duplicate_module_scores) > 0) {
    stop("Duplicated sample <U+00D7> module entries detected in SHAP bridge.")
}
shap_module_metadata <- bind_rows(ko_meta_shap %>% transmute(module_id, display_label, module_type, axis, row_color, row_order, n_features), met_meta_shap %>% transmute(module_id, display_label, module_type, axis, row_color, row_order, n_features), img_meta_shap %>% transmute(module_id, display_label, module_type, axis, row_color, row_order, n_features)) %>% mutate(module_uid = module_id, axis = as.character(axis), display_label = as.character(display_label), module_type = as.character(module_type), 
    row_color = as.character(row_color), row_order = as.numeric(row_order)) %>% arrange(row_order, module_id)
if (anyDuplicated(shap_module_metadata$module_id)) {
    stop("Duplicated module_id detected in SHAP metadata.")
}
shap_module_score_wide <- shap_module_scores_long %>% tidyr::pivot_wider(names_from = module_id, values_from = module_score) %>% arrange(match(sample, common_shap_samples))
shap_module_score_df <- as.data.frame(shap_module_score_wide, check.names = FALSE)
rownames(shap_module_score_df) <- shap_module_score_df$sample
shap_module_score_df$sample <- NULL
ordered_shap_modules <- shap_module_metadata$module_id[shap_module_metadata$module_id %in% colnames(shap_module_score_df)]
shap_module_score_df <- shap_module_score_df[, ordered_shap_modules, drop = FALSE]
shap_module_metadata <- shap_module_metadata %>% filter(module_id %in% colnames(shap_module_score_df)) %>% arrange(match(module_id, colnames(shap_module_score_df)))
if (!identical(colnames(shap_module_score_df), shap_module_metadata$module_id)) {
    stop("SHAP score matrix and metadata module order do not match.")
}
shap_numeric_mat <- as.matrix(shap_module_score_df)
storage.mode(shap_numeric_mat) <- "numeric"
module_finite_fraction <- colMeans(is.finite(shap_numeric_mat))
sample_finite_fraction <- rowMeans(is.finite(shap_numeric_mat))
if (any(module_finite_fraction <= 0)) {
    stop("At least one SHAP module contains no finite patient-level scores.")
}
shap_bridge_module_audit <- shap_module_metadata %>% mutate(finite_fraction = module_finite_fraction[module_id], n_finite = colSums(is.finite(shap_numeric_mat))[module_id])
shap_bridge_sample_audit <- data.frame(sample = rownames(shap_numeric_mat), finite_fraction = as.numeric(sample_finite_fraction), n_finite_modules = rowSums(is.finite(shap_numeric_mat)), n_total_modules = ncol(shap_numeric_mat), stringsAsFactors = FALSE)
score_matrix_file <- file.path(SHAP_BRIDGE_DIR, get_value("science.tables.attribution_042", "attribution_042.csv"))
metadata_file <- file.path(SHAP_BRIDGE_DIR, get_value("science.tables.attribution_043", "attribution_043.csv"))
module_audit_file <- file.path(SHAP_BRIDGE_DIR, get_value("science.tables.attribution_046", "attribution_046.csv"))
sample_audit_file <- file.path(SHAP_BRIDGE_DIR, get_value("science.tables.attribution_047", "attribution_047.csv"))
long_audit_file <- file.path(SHAP_BRIDGE_DIR, get_value("science.tables.attribution_048", "attribution_048.csv"))
write.csv(shap_module_score_df, score_matrix_file, row.names = TRUE, quote = TRUE)
write.csv(shap_module_metadata, metadata_file, row.names = FALSE, quote = TRUE)
write.csv(shap_bridge_module_audit, module_audit_file, row.names = FALSE, quote = TRUE)
write.csv(shap_bridge_sample_audit, sample_audit_file, row.names = FALSE, quote = TRUE)
write.csv(shap_module_scores_long, long_audit_file, row.names = FALSE, quote = TRUE)
```

```r
if (!requireNamespace("openxlsx", quietly = TRUE)) {
    install.packages("openxlsx")
}
library(openxlsx)
library(dplyr)
TABLE_DIR <- file.path(getwd(), get_value("science.paths.directory_107", "directory_107"))
dir.create(TABLE_DIR, recursive = TRUE, showWarnings = FALSE)
OUTPUT <- file.path(TABLE_DIR, get_value("science.tables.result_300", "result_300.xlsx"))
collect_tables <- function(keywords) {
    objs <- ls(envir = .GlobalEnv)
    res <- list()
    for (nm in objs) {
        obj <- get(nm, envir = .GlobalEnv)
        if (is.data.frame(obj)) {
            low <- tolower(nm)
            if (any(sapply(keywords, function(k) grepl(tolower(k), low, fixed = TRUE)))) {
                res[[nm]] <- obj
            }
        }
    }
    res
}
combine_tables <- function(x) {
    if (length(x) == 0) {
        return(NULL)
    }
    out <- bind_rows(lapply(names(x), function(nm) {
        x[[nm]] %>% mutate(source_object = nm)
    }))
    out
}
tables <- list()
axis_tbl <- collect_tables(c("axis", "ecosystem", "summary", "annotation", "module"))
tables$S9_Ecosystem_axis_annotation <- combine_tables(axis_tbl)
ko_tbl <- collect_tables(c("ko", "kegg", "orthology"))
tables$S9_KO_module_annotation <- combine_tables(ko_tbl)
met_tbl <- collect_tables(c("metabolite", "metab", "compound"))
tables$S9_Metabolite_module_annotation <- combine_tables(met_tbl)
rad_tbl <- collect_tables(c("radiomics", "rad", "phenotype"))
tables$S9_Radiomics_module_annotation <- combine_tables(rad_tbl)
tables <- tables[!sapply(tables, is.null)]
wb <- createWorkbook()
for (nm in names(tables)) {
    addWorksheet(wb, substr(nm, 1, 31))
    writeData(wb, sheet = substr(nm, 1, 31), tables[[nm]])
    addFilter(wb, sheet = substr(nm, 1, 31), rows = 1, cols = seq_len(ncol(tables[[nm]])))
    freezePane(wb, sheet = substr(nm, 1, 31), firstRow = TRUE)
    setColWidths(wb, sheet = substr(nm, 1, 31), cols = 1:ncol(tables[[nm]]), widths = "auto")
}
addWorksheet(wb, "Notes")
writeData(wb, "Notes", data.frame(Item = c("KO annotation", "Metabolite annotation", "Radiomics annotation", "Statistics"), Description = c("Raw KO identifiers and biological annotations retained.", "Biological module labels are display annotations only.", "Imaging phenotype modules are exported without changing original features.", "No statistical recalculation is performed during table export.")))
saveWorkbook(wb, OUTPUT, overwrite = TRUE)
for (nm in names(tables)) {
}
```
